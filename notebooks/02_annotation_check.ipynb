# Annotation Validation

Check and validate annotations across all datasets.

## 1. Setup

```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
import cv2

sys.path.insert(0, str(Path.cwd() / 'src'))

from datasets.validate_yolo_dataset import validate_yolo_dataset_structure, count_classes_in_labels

sns.set_style('whitegrid')
```

## 2. Validate EASY-v0 Structure

```python
# Check if EASY-v0 exists and validate
easy_v0_path = Path('data/processed/EASY-v0')

if easy_v0_path.exists():
    is_valid, errors = validate_yolo_dataset_structure(str(easy_v0_path))
    
    if is_valid:
        print("✓ EASY-v0 dataset structure is valid")
    else:
        print("✗ Validation errors found:")
        for error in errors:
            print(f"  - {error}")
else:
    print(f"Dataset not found at {easy_v0_path}")
```

## 3. Analyze Label Distribution

```python
labels_dir = easy_v0_path / 'labels' / 'train'

if labels_dir.exists():
    class_counts = count_classes_in_labels(str(labels_dir))
    
    if class_counts:
        df = pd.DataFrame([
            {'Class ID': cid, 'Count': count}
            for cid, count in sorted(class_counts.items())
        ])
        
        class_names = {
            0: 'ship', 1: 'boat', 2: 'speedboat',
            3: 'sailboat', 4: 'buoy', 5: 'structure',
            6: 'helicopter', 7: 'aircraft'
        }
        df['Class Name'] = df['Class ID'].map(class_names)
        
        print(df.to_string(index=False))
    else:
        print("No labels found")
else:
    print(f"Labels directory not found: {labels_dir}")
```
