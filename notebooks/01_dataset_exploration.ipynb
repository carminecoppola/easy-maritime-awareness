# Dataset Exploration

Explore the maritime datasets from public sources and analyze their structure, statistics, and characteristics.

## 1. Import Libraries

```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict

# Add src to path
sys.path.insert(0, str(Path.cwd() / 'src'))

from datasets.inspect_dataset import count_images_by_source, inspect_annotations

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 2. Analyze Dataset Sources

```python
# Count images by source
raw_data_dir = Path('data/raw')
image_counts = count_images_by_source(str(raw_data_dir))

print("Image counts by source:")
for source, count in sorted(image_counts.items()):
    print(f"  {source}: {count}")

total = sum(image_counts.values())
print(f"\nTotal images: {total}")
```

## 3. Visualize Dataset Statistics

```python
if image_counts:
    fig, ax = plt.subplots(figsize=(12, 6))
    sources = list(image_counts.keys())
    counts = list(image_counts.values())
    
    ax.bar(sources, counts, color='steelblue')
    ax.set_xlabel('Dataset Source')
    ax.set_ylabel('Number of Images')
    ax.set_title('Dataset Distribution by Source')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
else:
    print("No images found in data/raw directory")
```
